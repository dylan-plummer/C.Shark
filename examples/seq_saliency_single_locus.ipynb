```python
import subprocess
import pandas as pd
from tqdm import tqdm

models = pd.read_csv('tmp/64_models.tsv', sep='\t', names=['path', 'targets'])

res = {'model': [],
       'motif': [],
       'pos_100bp': [],
       'pos': [],
       'score_fw': [],
       'score_rc': [],
       'corr_max': [],
       'corr_min': [],
       'mean_hic_change': [],
       'target': []}
for model_i, row in tqdm(models.iterrows(), total=models.shape[0]):
    model_path = row['path']
    model_targets = row['targets'].split(',')
    model_targets = [t.strip() for t in model_targets]
    if 'no_training' in model_targets:
        continue
    # run the saliency score script for each model
    cmd = [
        'python', 'seq_saliency_scores.py',
        '--model', model_path,
        '--chr', 'chr11',
        '--start', '31000000',
        '--seq', '../cshark_data/data/hg19/dna_sequence',
        '--bigwigs', 'ctcf=../cshark_data/data/hg19/hESC_WT_5kb_norm_10/genomic_features/ctcf.bw',
                    'atac=../cshark_data/data/hg19/hESC_WT_5kb_norm_10/genomic_features/atac.bw',
        '--target-locus', 'chr11:31760000-31860000_chr11:32380000-32480000',
        '--plot-locus', 'chr11:31373941-33024355',
        '--resolution', '4096',
        '--matrix-size', '512',
        '--out-dir', 'tmp',
        '--ctcf-ko',
        '--n-motifs', '10'
    ]
    subprocess.run(cmd)
    # read output file and map motifs to hic diffs
    df = pd.read_csv('tmp/motif_insertion_effects_summary.tsv', sep='\t')
    for _, r in df.iterrows():
        motif = r['motif']
        res['model'].append(model_path)
        res['motif'].append(motif)
        res['pos_100bp'].append(r['pos_100bp'])
        res['pos'].append(r['pos'])
        res['score_fw'].append(r['score_fw'])
        res['score_rc'].append(r['score_rc'])
        res['corr_max'].append(r['corr_max'])
        res['corr_min'].append(r['corr_min'])
        res['mean_hic_change'].append(r['mean_hic_change'])
        res['target'].append(row['targets'])
    result_df = pd.DataFrame(res)
    result_df.to_csv('tmp/combined_motif_saliency_scores.tsv', sep='\t', index=False)
    
```
